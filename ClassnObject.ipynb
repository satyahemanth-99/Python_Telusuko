Polymorphism

```python
def mul(*args):

  total = 1
  for i in args:
    total *= i

  return total
```

```python
mul(2, 6, 4, 3, 7)
```

```python
class Human:

  def speak(self, Language):
    print(f"I can speak {Language}")
```

```python
h1 = Human()
h2 = Human()

h1.speak("English")
h2.speak("Hindi")
```
