```python
class Kids:
  def __init__(self):
    self.name = "Viyaan"
    self.age = 2

  def compare(self, other):
    if self.age == other.age:
      return True
    else:
      return False
```

```python
leo = Kids()
ivaan = Kids()
leo.age = 1
```

```python
if leo.compare(ivaan):
  print("They are same")
else:
  print("They are different")
```

```python
class Student:

  school = "Telusko"

  def __init__(self, m1, m2, m3):
    self.m1 = m1
    self.m2 = m2
    self.m3 = m3

  def avg(self):
    return (self.m1 + self.m2 + self.m3)/3

  def get_m1(self):
    return self.m1

  def set_m1(self, value):
    self.m1 = value

  @classmethod
  def info(cls):
    return cls.school
```

```python
s1 = Student(34, 67, 32)
s2 = Student(89, 2, 12)

s1.avg(), s2.avg(), s1.school, s2.school
```
